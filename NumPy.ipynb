```python
from IPython import get_ipython
from IPython.display import display
# %% [markdown]
# # NumPy
# 
# 
# %% [markdown]
# ## 1. Importing NumPy
# 
# **Explanation:** NumPy is imported with the common alias np. The exact version depends on the Colab runtime.
# 
# %%
import numpy as np



# %% [markdown]
# ## 2. Creating a 1-D Array
# 
# **Explanation:** np.array() converts a Python sequence into a NumPy ndarray.
# 
# %%
import numpy as np

# Create a one-dimensional NumPy array
n=np.array([1,2,3,4])




# %% [markdown]
# ## 3. Checking Type
# 
# **Explanation:** A NumPy array is an object of type numpy.ndarray.
# 
# 
# %%
a = np.array([10, 20, 30])
b=np.array((10,14,13))
# Check the object type
print(type(a))
print(a.ndim)
print(type(b))
# %% [markdown]
# ## 4. dtype
# 
# **Explanation:** dtype tells us the data type stored in the array.
# 
# 
# %%
a = np.array([10, 20, 30])

# Check the data type of elements
print(a.dtype)


# %% [markdown]
# ## 5. ndim
# 
# **Explanation:** ndim tells how many dimensions the array has.
# 
# 
# %%
a = np.array([[[1, 2, 3], [4, 5, 6]]])

# Number of dimensions
print(a.ndim)

# %% [markdown]
# ## 6. shape
# 
# **Explanation:** (2, 3) means 2 rows and 3 columns.
# 
# 
# %%
a = np.array([[1, 2, 3], [4, 5, 6]])

# Check rows and columns
print(a.shape)

# %% [markdown]
# ## 7. size
# 
# **Explanation:** size is the total number of elements.
# 
# 
# %%
a = np.array([[1, 2, 3], [4, 5, 6]])

# Total number of elements
print(a.size)


# %% [markdown]
# ## 8. zeros()
# 
# **Explanation:** zeros() creates an array filled with zeros.
# 
# 
# %%
# Create five zeros
zero=np.zeros(12,dtype=int)
print(zero)
# %% [markdown]
# ## 9. zeros() with Shape
# 
# **Explanation:** A shape tuple such as (2, 3) creates 2 rows and 3 columns.
# 
# 
# %%
a = np.zeros((2, 3), dtype=int)

# Create a 2 x 3 integer matrix of zeros
print(a)

# %% [markdown]
# ## 10. ones()
# 
# **Explanation:** ones() creates an array filled with ones.
# 
# %%
# Create a 2 x 3 matrix of ones
a = np.ones((2,3,3), dtype=int)
print(a)

# %% [markdown]
# ## 11. full()
# 
# **Explanation:** full() fills every position with the specified value.
# 
# %%
a = np.full((4,2, 3), 9)

# Fill every position with 7
print(a)

# %% [markdown]
# ## 12. Identity Matrix
# 
# **Explanation:** eye() creates a matrix with 1s on the main diagonal.
# 
# 
# %%
a = np.eye(4, dtype=int)

# Create a 4 x 4 identity matrix
print(a)

# %% [markdown]
# ## 13. arange()
# 
# **Explanation:** arange() is step-based and excludes the stop value.
# 
# **Expected output:** Run the code cell below and compare the result.
# %%
# start=1, stop=11 (excluded), step=2
arr=np.arange(1,11,2)
print(arr)

# %%
arr = np.arange(2,51,2)
print(arr)
# %% [markdown]
# ## 14. linspace()
# 
# **Explanation:** linspace() is count-based and includes the end point by default.
# 
# %%
a = np.linspace(0,1 , 5 )

# Generate 5 equally spaced values from 0 to 1
print(a)


# %% [markdown]
# ## 15. Integer Indexing
# 
# **Explanation:** Use square brackets to access elements. Negative indices count from the end.
# 
# 
# %%
a = np.array([10, 20, 30, 40])

# Indexing starts at 0
print(a[1])
print(a[-1])

# -1 means the last element


# %% [markdown]
# ## 16. 2-D Indexing
# 
# **Explanation:** a[1, 2] means second row and third column.
# 
# 
# %%
a = np.array([[10, 20, 30], [40, 50, 60]])

# a[row, column]
print(a[1,1])
print(a[0][2])

# %% [markdown]
# ## 17. Selecting a Row
# 
# **Explanation:** The colon means select all columns.
# 
# 
# %%
a = np.array([[10, 20, 30], [40, 50, 60]])

# Select all columns from row index 1
print(a[1, :])

# %% [markdown]
# ## 18. Selecting a Column
# 
# **Explanation:** The colon selects all rows.
# 
# 
# %%
a = np.array([[10, 20, 30], [40, 50, 60]])
print(a[:,1])

# Select all rows from column index 1


# %% [markdown]
# ## 19. Basic Slicing
# 
# **Explanation:** The start is included and the stop is excluded.
# 
# 
# %%
a = np.arange(1, 11)
print(a)
# Select indices 2 through 6; stop index 7 is excluded
print(a[2:7])
print(a[1:])
print(a[:])
print(a[-6:-2])
print(a[-2:-6])

# %% [markdown]
# ## 20. Step Slicing
# 
# **Explanation:** A step of 2 selects every second element.
# 
# 
# %%
a = np.arange(1, 11)

# Select every second element
print(a[::2])

# %% [markdown]
# ## 21. Reverse an Array
# 
# **Explanation:** Negative step values traverse an array backwards.
# 
# 
# %%
a = np.arange(1, 6)
print(a)
# A step of -1 reverses the array
print(a[::-2])
print(a.ndim)

# %% [markdown]
# ## 22. 2-D Slicing
# 
# **Explanation:** 2-D slicing uses array[row_slice, column_slice].
# 
# 
# %%
a = np.arange(1, 13).reshape(2,3,2)

# Select rows 0-1 and columns 1-2
#print(a[0:2, 1:3])
print(a)
print(a.ndim)
# %% [markdown]
# ## 23. reshape()
# 
# **Explanation:** reshape() changes the shape while preserving the total number of elements.
# 
# 
# %%
a = np.arange(1, 13).reshape(3,4)
print(a)
# Reshape 12 elements into 3 rows and 4 columns
# %% [markdown]
# ## 24. Reshaping 30 Elements
# 
# **Explanation:** A reshape is valid only when the product of the new dimensions equals the number of elements.
# 
# 
# %%
# Reshape 30 elements into 3 D array with 5 layers
a= np.arange(1,31).reshape(5,2,3)
print(a)




# %% [markdown]
# ## 25. flatten()
# 
# **Explanation:** flatten() produces a 1-D copy.
# 
# 
# %%
a = np.array([[1, 2], [3, 4]])
print(a.flatten())
# Convert to 1-D; flatten returns a copy


# %% [markdown]
# ## 26. Transpose
# 
# **Explanation:** Transpose converts rows into columns.
# 
# 
# %%
a = np.array([[1, 2, 3], [4, 5, 6]])

# Swap rows and columns
print(a.T)


# %% [markdown]
# ## 27. Element-wise Addition
# 
# **Explanation:** NumPy performs element-wise addition.
# 
# 
# %%
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

# Add corresponding elements
print(a * b)

# %% [markdown]
# ## 28. Multiplication
# 
# **Explanation:** Scalar operations are automatically applied to every element.
# 
# **Expected output:** Run the code cell below and compare the result.
# %%
a = np.array([1, 2, 3])

# Multiply every element by 5
print(a * 5)

# %% [markdown]
# ## 29. Comparison
# 
# **Explanation:** Comparisons produce a Boolean array.
# 
# 
# %%
a = np.array([10, 20, 30, 40])

# Compare every element with 25
print(a > 25)

# %% [markdown]
# ## 30. sum()
# 
# **Explanation:** sum() adds all elements.
# 
# 
# %%
a = np.array([10, 20, 30, 40])

# Add all elements
print(a.sum())

# %% [markdown]
# ## 31. sum(axis=0) and axis=1
# 
# **Explanation:** For a 2-D array, axis=0 aggregates down rows and axis=1 aggregates across columns.
# 
# **Expected output:** Run the code cell below and compare the result.
# %%
a = np.array([[1, 2, 3], [4, 5, 6]])

# axis=0 -> column-wise aggregation
print(a.sum(axis=0))
# axis=1 -> row-wise aggregation
print(a.sum(axis=1)) # by default

# %% [markdown]
# ## 32. mean()
# 
# **Explanation:** Mean = sum of values divided by the number of values.
# 
# 
# %%
a = np.array([10, 20, 30, 40])

# Calculate the arithmetic mean
print(a.mean())



# %% [markdown]
# ## 33. min(), max()
# 
# **Explanation:** min() returns the smallest value and max() returns the largest.
# 
# %%
a = np.array([12, 5, 27, 9])

# Find smallest and largest values
print(a.min())
print(a.max())


# %% [markdown]
# ## 37. std() and var()
# 
# **Explanation:** Standard deviation is the square root of variance.
# 
# 
# %%
a = np.array([2, 4, 6, 8, 10])

# Variance and standard deviation
print(a.std())
print(a.var())

# %% [markdown]
# ## 38. argmax() and argmin()
# 
# **Explanation:** argmax() and argmin() return positions, not the values themselves.
# 
# 
# %%
a = np.array([10, 50, 20, 80, 30])

# Find indices of maximum and minimum values

print(a.argmax())
print(a.argmin())
# %% [markdown]
# ## 39. Boolean Filtering
# 
# **Explanation:** Boolean indexing filters an array using a condition.
# 
# %%
a = np.array([5, 12, 7, 20, 3, 18])

# Keep values greater than 10
print(a[a > 10])


# %% [markdown]
# ## 40. Even Numbers
# 
# **Explanation:** The modulo operator (%) checks the remainder.
# 
# 
# %%
a = np.arange(1, 11)

# Select values divisible by 2
print(a[a % 2 == 0])
print(a[a%2!=0])

# %% [markdown]
# ## 41. Multiple Conditions
# 
# **Explanation:** Use & for element-wise AND and put each condition in parentheses.
# 
# 
# %%
a = np.arange(1, 11)

# Select values greater than 3 AND less than 8
print(a[(a > 3) & (a < 8)])

# %% [markdown]
# ## 42. np.where() as a Condition
# 
# **Explanation:** np.where(condition, true_value, false_value) performs vectorized conditional selection.
# 
# 
# %%
marks = np.array([35, 67, 42, 28, 80])

# Label marks as Pass or Fail
result = np.where(marks >= 40,'true','false')
print(result)

# %% [markdown]
# ## 43. np.where() to Find Positions
# 
# **Explanation:** With only a condition, np.where() returns matching indices.
# 
# 
# %%
a = np.array([5, 12, 7, 20, 3, 18])

# Find indices where values are greater than 10
print(np.where(a > 10))

# %% [markdown]
# ## 44. Sorting
# 
# **Explanation:** sort() returns values in ascending order.
# 
# 
# %%
a = np.array([40, 10, 30, 20])

# Sort in ascending order
print(np.sort(a))

# %% [markdown]
# ## 46. unique()
# 
# **Explanation:** unique() returns sorted unique values.
# 
# 
# %%
a = np.array([2, 3, 2, 5, 3, 2])

# Remove duplicates
print(np.unique(a))

# %% [markdown]
# ## 48. concatenate()
# 
# **Explanation:** concatenate() joins arrays along an existing axis.
# 
# 
# %%
a = np.array([1, 5, 3])
b = np.array([4, 2, 6])

# Join arrays end-to-end
print(np.concatenate((a,b)))

# %% [markdown]
# ## 51. split()
# 
# **Explanation:** split() divides an array into equal sections when possible.
# 
# 
# %%
a = np.arange(12)

# Split 12 elements into 3 equal arrays
print(np.split(a, 3))

# %% [markdown]
# ## 56. Matrix Multiplication
# 
# **Explanation:** The @ operator performs matrix multiplication.
# 
# 
# %%
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# @ performs matrix multiplication
print(A @ B)

# %% [markdown]
# ## 57. Element-wise vs Matrix Multiplication
# 
# **Explanation:** This is an important distinction: * multiplies corresponding elements, while @ follows matrix multiplication rules.
# 
# 
# %%
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# * = element-wise multiplication.    *-> element wise
#. @--> matrix
print(A * B)

# @ = matrix multiplication
print(A @ B)

# %% [markdown]
# ## 67. Dot Product
# 
# **Explanation:** The dot product is 1×4 + 2×5 + 3×6 = 32.
# The dot product is the sum of the products of corresponding elements.
# 
# %%
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Calculate the dot product
print(np.dot(a, b))

# %% [markdown]
# ## 70. Final
# 
# 
# 
# %%
marks = np.array([[80, 70, 90],
                  [35, 45, 40],
                  [95, 88, 92]])

# Calculate average marks for every student
avg = marks.mean(axis=1,)

# Classify students using a pass threshold
status = np.where(avg >= 40, 'Pass', 'Fail')

print('Average:', avg)
print('Status:', status)

```
