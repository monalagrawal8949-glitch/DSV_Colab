```python
from IPython import get_ipython
from IPython.display import display
# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# %%
#Sampling
#1. Need to create our dataset
np.random.seed(42)
df = pd.DataFrame({
    'Student_ID' :  range(1,101),
    'Name': [f'Student_{i}' for i in range (1,101)],
    'Gender' :  np.random.choice(['Male','Female'],size = 100),#100 bhi likh skte hai
    'Department':np.random.choice(['CSE','ECE','ME','CE'],size = 100),
    'Marks':np.random.randint(40 , 101 , 100),

})
# %%
#population length
print("POPULATION LENGTH" , len(df))
# %%
df.head()
# %%
#Department value?
print("Department Distribution Value" , df['Department'].value_counts())
#Gender Distribution
print("Gender Distribution Value" , df['Gender'].value_counts())
#Marks Distribution
print("Marks Distribution Value" , df['Marks'].value_counts())

# %%
#1.Simple random sampling
#every population(student) has an equal opportunity to be selected
sample_size = 10
simple_random_sample = df.sample(
n= sample_size ,
random_state = 42
)
simple_random_sample
# %%
#Visualization of SRS
plt.figure(figsize = (12,5))

plt.scatter(df['Student_ID'] , df['Marks'], alpha = 0.35 , label = 'Population')
plt.scatter(simple_random_sample['Student_ID'] ,
            simple_random_sample['Marks'], s=120 ,
            label = 'Simple Random Sample')
plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.title("----Simple Random Sampling----")

plt.legend()
plt.show()
# %%
#SYSTEMATIC SAMPLING:
N = len(df)
n =10
k=N//n
start = np.random.randint(0,k)
systematic_Sample = df.iloc[start::k].head(n)
systematic_Sample

print("\nStarting Position:" , start)
print("\nSampling Interval:" , k)
systematic_Sample
# %%
#Visualization of systematic sampling
plt.figure(figsize = (12,5))

plt.scatter(df['Student_ID'] , df['Marks'], alpha = 0.55 , label = 'Population')
plt.scatter(simple_random_sample['Student_ID'] ,
            simple_random_sample['Marks'], s=120 ,
            label = 'Selected Sample')
plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.title("----systematic_Sample---")

plt.legend()
plt.show()
# %%
from sklearn.model_selection import train_test_split
# %%
#3. Stratified Sampling
sample_size = 20
stratified_sample , remaining = train_test_split(
    df , test_size = 0.80,
    stratify = df['Department'],
    random_state = 42

)
stratified_sample

# %%
print("Original Proportions:")
print(df['Department'].value_counts(normalize = True).round(2))
print("\n Sample Distribution Proportions:")
print(stratified_sample['Department'].value_counts(normalize = True).round(2))
# %%
#Visualization of Stratified Sampling

comparison = pd.DataFrame({
    'population' : df['Department'].value_counts(),
    'Sample' : stratified_sample['Department'].value_counts()
})
comparison.plot(kind = 'bar' , figsize = (5,2))
plt.xlabel("Department")
plt.ylabel("MARKS")
plt.title("----Population Vs Stratified Sample--")
#plt.xticks(rotation=1)
plt.show()
# %%

```
