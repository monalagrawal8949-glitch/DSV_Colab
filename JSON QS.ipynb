```python
from IPython import get_ipython
from IPython.display import display
# %% [markdown]
# JSON BASICS
# 
# 
# 
# %%
#use this JSON text

t = ('{"shop": "Fresh Mart" , "open" :  true , "owner" : null, "items": ['
   '{"name" : "Rice" , "price" : 60 , "qty" :10}, '
   '{"name": "Oil" , "price" :150 , "qty" :4},'
   '{"name" : "Tea" , "price" : 120 , "qty" : 6}]}')

# %%
#load  t into a python object and print its type 
import json

data = json.loads(t)

print(type(data))
# %%
#print the name of the second item and the number of items
print(data["items"][1]["name"])
print(len(data["items"]))
# %%
#print the value of owner.What Python value is it?
print(data["owner"])
print(type(data["owner"]))
# %%
#convert the object back to JSON text with indentation 2
import json

json_text = json.dumps(data, indent=2)

print(json_text)
# %%

```
