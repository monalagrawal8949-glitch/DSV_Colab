```python
from IPython import get_ipython
from IPython.display import display
# %%
#create a numpy array from[10,20,30,40,50]. Display its size , shape and data type
import numpy as np
a = np.array([10,20,30,40,50])
print(a);

#shape
print("SHAPE:" , a.shape)
#size
print("SIZE:", a.size)
#data type
print("DATATYPE" , a.dtype)
# %%
# Create  a 3*4 array containing numbers from 1 to 12
a = np.arange(1,13).reshape(3,4)
print(a)

# %%
#create an zero matrix , identity matrix , ones matrix
#zero matrix

matrix = np.zeros((3,3),dtype=int)
print(matrix)

#ones matrix
matrix = np.ones((3,3))
print(matrix)

#identity matrix
matrix =  np.eye(3)
print(matrix)
# %%
# generate all even numbers between 10 and 50 using np.arange()
even_numbers = np.arange(10,51,2)
print(even_numbers)
# %%
#create an array of 10 equally spaced numbers between 0 and 1 using np.linspace()
import numpy as np
arr = np.linspace(0,1,10)
print(arr)
# %% [markdown]
# Array arithmetic
# %%
#given two numpy arrays perfrom element wise addition subtraction multiplication and division
a =  np.array([10,20,30, 40])
b =  np.array([2,4,5,8])


print(a+b)
print(a-b)
print(a*b)
print(a/b)
# %%
#calculate the square and cube of every element in a Numpy array
import numpy as np
arr = np.array([1,2,3,4,5])
print(np.square(arr))
print(np.power(arr,3))
# %%
#increase every element of array by 15%
arr = np.array([100 , 200 , 300 , 400])
new_arr = arr*1.15
print(new_arr)
# %%
#given two element find their element wise maximum and minimum
a = np.array([1,2,3,])
b= np.array([4,0,9])
print(np.maximum(a,b))
print(np.minimum(a,b))

# %% [markdown]
# indexing or slicing
# 
# 
# %%
#create an array from 0 to 20 and extract the first five elements , last five elements , elements from index 5 to 15 , and every second element
arr = np.arange(21)

first_five = arr[:5]
print(first_five)

last_five = arr[-5:]
print(last_five)

middle_elements = arr[5:16]
print(middle_elements)

second_elements = arr[::2]
print(second_elements)
# %%
#replace the first five elements of an array with 100
arr =  np.array([10,20,30,40,50,60,60,70])
arr[:5] = 100
print(arr)
# %%
#create an array and demonstrate the difference between a slice/view and a copy
arr = np.array([10,20,30, 40,50,60,70,80])

view = arr[1:4]
view[0] = 100

print("original after view" , arr)
print("view" , view)

#copy
copy = arr[1:4].copy()
copy[0] = 200

print("original after copy" , arr)
print("copy" , copy)
# %%

```
