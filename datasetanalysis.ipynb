```python
from IPython import get_ipython
from IPython.display import display
# %%
import pandas as pd
import numpy as np
# %%

# %% [markdown]
# 
# %%
data = pd.read_csv('/content/Zomato Restaurant Dataset (1).csv')
df = pd.DataFrame(data)
print(df.isnull().sum())
# %%
import pandas as pd
df = pd.read_csv('/content/diabetes.csv')
# %%
df.isna()
# %%
df.isna().sum()
# %%
df = df.replace(0 , np.nan)
# %%
df.isna().sum()
# %%
import pandas as pd
import numpy as np


columns = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']


df = pd.read_csv('/content/diabetes.csv')
column_name = 'BMI'

Q1 = df[column_name].quantile(0.25)
Q3 = df[column_name].quantile(0.75)
IQR = Q3 - Q1

print(f"--- {column_name} ke liye IQR ---")
print(f"Q1 (25th Percentile): {Q1}")
print(f"Q3 (75th Percentile): {Q3}")
print(f"IQR Value: {IQR}\n")


lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR


df_clean = df[(df[column_name] >= lower_bound) & (df[column_name] <= upper_bound)]

print(f"Purana Data Shape: {df.shape}")
print(f"Outliers hatane ke baad Naya Data Shape: {df_clean.shape}")
# %%
# BMI ke liye Mean, Median aur Mode nikalna
bmi_mean = df_clean['BMI'].mean()
bmi_median = df_clean['BMI'].median()
bmi_mode = df_clean['BMI'].mode()[0]

print(f"--- BMI ke Statistics (Outliers hatane ke baad) ---")
print(f"Mean   : {bmi_mean:.2f}")
print(f"Median : {bmi_median:.2f}")
print(f"Mode   : {bmi_mode:.2f}\n")

# Poore dataset ke sabhi numeric columns ka Mean aur Median ek sath dekhne ke liye
print("--- Sabhi Columns ka Mean aur Median ---")
stats_df = pd.DataFrame({
    'Mean': df_clean.mean(),
    'Median': df_clean.median()
})
display(stats_df)
```
